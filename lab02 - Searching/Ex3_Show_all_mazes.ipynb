## Thống kê thông số 8 mê cung ($n, d, m, b$)

Bảng tổng hợp thông số cho 8 bản đồ mê cung:

| :--- | :---: | :---: | :---: | :---: | :---: | :---: | :---: |

### Giải thích các đại lượng:
- **$n$ (Số trạng thái / Số ô đi được):** Tổng số ô không chứa tường `X` trên mê cung (bao gồm ô xuất phát $S$, ô đích $G$, và các ô đường đi ` `).
- **$d$ (Độ sâu lời giải tối ưu):** Số bước ngắn nhất (cost / depth) đi từ vị trí $S$ tới vị trí $G$ (thu được qua thuật toán BFS / A*).
- **$m$ (Độ sâu tối đa):** Chiều dài tối đa của đường đi không lặp lại ô ($m = n$).
- **$b$ (Hệ số phân nhánh tối đa):** Số lượng hướng đi hợp lệ tối đa (Lên, Xuống, Trái, Phải) từ bất kỳ ô đi được nào trên bản đồ ($b \le 4$).
